```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Step 1: Load the dataset
data = pd.read_csv("/content/student_scores.csv")

# Step 2: Visualize the data
plt.scatter(data['Hours'], data['Scores'], color='blue')
plt.title('Study Hours vs Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Score (%)')
plt.grid(True)
plt.show()

# Step 3: Prepare the data
X = data[['Hours']]   # Input features (2D)
y = data['Scores']    # Target (1D)

# Step 4: Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Step 5: Train the model
model = LinearRegression()
model.fit(X_train, y_train)

# Step 6: Make predictions
predictions = model.predict(X_test)

# Step 7: Plot the regression line
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color='red')  # Regression line
plt.title('Regression Line: Study Hours vs Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Score (%)')
plt.grid(True)
plt.show()

# Step 8: Predict for a custom input
hours = float(input("Enter hours studied: "))
predicted_score = model.predict([[hours]])
print(f"Predicted Score: {predicted_score[0]:.2f}%")
```
